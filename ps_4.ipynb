```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./data/heart_failure_clinical_records_dataset.csv')
df
```

**a. Plot a KDE of ejection_fraction:**

```python
sns.kdeplot(x=df['ejection_fraction'])
```

**b. Use Pandas' df.sample(frac=1.0, replace=True) to resample the data 15 times with replacement, plotting the kernel density estimtes for each sample**

```python
for i in range(15):
    samplei = df.sample(frac=01.0, replace=True)
    sns.kdeplot(x=samplei['ejection_fraction'])

sns.kdeplot(x=df['ejection_fraction'], linestyle='--', color='black')
plt.show()
```

**c. For what values of ejection_fraction is the original plot reliable? For which values is there noticeable variation in your plots of the resampled data?**

I plotted the original plot with a black dashed line and overlaid all of the resampled KDEs. From 0 to around 20 the original plot seems to be reliable as well as for ejection fractions roughly between 42-48. From 65 onward they also seem reliable. That is because there is less variation in the resampled lines in these ejection fraction regions. However, at crests and bumps in the KDE there is much more variation: notably when ejection fraction roughly equals 25, 35, 55 and 60.
